```python
import os
import pickle
import numpy as np
import pandas as pd

os.makedirs(
    "../Outputs/Notebook_14",
    exist_ok=True
)

print(
    "Libraries Loaded Successfully"
)
```

# Load Encrypted Templates

```python
with open(
    "../Outputs/Notebook_06/encrypted_embeddings.pkl",
    "rb"
) as f:

    encrypted_embeddings = pickle.load(f)

print(
    "Encrypted Templates Loaded"
)

print(
    "Total Templates:",
    len(encrypted_embeddings)
)
```

# Simulate Template Theft Attack

```python
stolen_template = encrypted_embeddings[0]

print(
    "Template Retrieved By Attacker"
)

print(
    "Data Type:",
    type(stolen_template)
)

print(
    "Ciphertext Size:",
    len(stolen_template),
    "bytes"
)
```

# Analyze Ciphertext Structure

```python
print(
    "First 100 Ciphertext Bytes:"
)

print(
    stolen_template[:100]
)
```

# Attempt Feature Reconstruction

```python
try:

    reconstructed_feature = np.array(
        stolen_template,
        dtype=np.float64
    )

    print(
        "Feature Reconstruction Successful"
    )

except Exception:

    print(
        "Feature Reconstruction Failed"
    )
```

# Attempt Image Reconstruction

```python
try:

    image_data = np.array(
        stolen_template
    )

    image_data.reshape(
        224,
        224
    )

    print(
        "Image Reconstruction Successful"
    )

except Exception:

    print(
        "Image Reconstruction Failed"
    )
```
