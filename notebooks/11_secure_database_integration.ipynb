```python
import os
import pickle
import pandas as pd
import tenseal as ts

os.makedirs(
    "../Outputs/Notebook_11",
    exist_ok=True
)

print(
    "Libraries Loaded Successfully"
)
```

# Load Encryption Context

```python
with open(
    "../Outputs/Notebook_06/context.pkl",
    "rb"
) as f:

    context_bytes = pickle.load(f)

context = ts.context_from(
    context_bytes
)

print(
    "Encryption Context Loaded"
)
```

# Load Encrypted Templates

```python
with open(
    "../Outputs/Notebook_06/encrypted_embeddings.pkl",
    "rb"
) as f:

    encrypted_embeddings = pickle.load(f)

print(
    "Encrypted Templates Loaded"
)

print(
    "Total Templates:",
    len(encrypted_embeddings)
)
```

# Create Secure Database

```python
secure_database = {}

for idx, template in enumerate(
    encrypted_embeddings
):

    secure_database[idx] = template

print(
    "Secure Database Created"
)

print(
    "Database Size:",
    len(secure_database)
)
```

# Retrieve Encrypted Template

```python
template_id = 0

retrieved_template = secure_database[
    template_id
]

print(
    "Template Retrieved Successfully"
)

print(
    type(
        retrieved_template
    )
)
```

# Verify Template Integrity

```python
original_template = encrypted_embeddings[
    template_id
]

if original_template == retrieved_template:

    print(
        "Integrity Verification Passed"
    )

else:

    print(
        "Integrity Verification Failed"
    )
```

# Database Statistics

```python
database_stats = pd.DataFrame({

    "Metric": [

        "Total Templates",
        "Stored Templates"

    ],

    "Value": [

        len(encrypted_embeddings),
        len(secure_database)

    ]

})

database_stats
```

```python
database_stats.to_csv(

    "../Outputs/Notebook_11/database_statistics.csv",

    index=False

)

print(
    "Database Statistics Saved"
)
```
