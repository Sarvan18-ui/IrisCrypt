```python
import os
import time
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import tenseal as ts

os.makedirs(
    "../Outputs/Notebook_13",
    exist_ok=True
)

print(
    "Libraries Loaded Successfully"
)
```

# Load Encryption Statistics

```python
encryption_stats = pd.read_csv(
    "../Outputs/Notebook_06/encryption_statistics.csv"
)

encryption_stats.head()
```

# Encryption Time Analysis

```python
avg_encryption_time = np.mean(
    encryption_stats["Encryption_Time"]
)

max_encryption_time = np.max(
    encryption_stats["Encryption_Time"]
)

min_encryption_time = np.min(
    encryption_stats["Encryption_Time"]
)

print(
    "Average Encryption Time:",
    avg_encryption_time
)

print(
    "Maximum Encryption Time:",
    max_encryption_time
)

print(
    "Minimum Encryption Time:",
    min_encryption_time
)
```

# Encryption Time Distribution

```python
plt.figure(
    figsize=(8,5)
)

plt.hist(

    encryption_stats["Encryption_Time"],

    bins=30

)

plt.title(
    "Encryption Time Distribution"
)

plt.xlabel(
    "Time (seconds)"
)

plt.ylabel(
    "Frequency"
)

plt.savefig(

    "../Outputs/Notebook_13/encryption_time_distribution.png"

)

plt.show()
```

# Load Memory Usage Statistics

```python
memory_stats = pd.read_csv(
    "../Outputs/Notebook_06/memory_usage_analysis.csv"
)

memory_stats.head()
```

# Memory Overhead Analysis

```python
avg_memory = np.mean(
    memory_stats["Memory_Bytes"]
)

max_memory = np.max(
    memory_stats["Memory_Bytes"]
)

min_memory = np.min(
    memory_stats["Memory_Bytes"]
)

print(
    "Average Memory Usage:",
    avg_memory,
    "Bytes"
)

print(
    "Maximum Memory Usage:",
    max_memory,
    "Bytes"
)

print(
    "Minimum Memory Usage:",
    min_memory,
    "Bytes"
)
```

# Memory Usage Distribution

```python
plt.figure(
    figsize=(8,5)
)

plt.hist(

    memory_stats["Memory_Bytes"],

    bins=30

)

plt.title(
    "Memory Usage Distribution"
)

plt.xlabel(
    "Memory (Bytes)"
)

plt.ylabel(
    "Frequency"
)

plt.savefig(

    "../Outputs/Notebook_13/memory_usage_distribution.png"

)

plt.show()
```
